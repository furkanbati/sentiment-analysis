```python
import pandas as pd
import re
from sklearn.feature_extraction.text import TfidfVectorizer
# Load the data
# First 10k row in order to speed up the learning process
df = pd.read_csv('datasets\IMDB Dataset.csv')[:10000]
df.head()
```

```python
def clean(text):
    # Delete the HTML tags
    text = re.sub(r'<.*?>', '', text)
    # Just keep the letters, delete the numbers and marks
    text = re.sub(r'[^a-zA-Z]', ' ', text)
    # Convert to small letters
    return text.lower()
```

```python
df["review"] = df['review'].apply(clean)
```

Looks like we succesfully cleaned the reveiew text

Now lets veectorize the words using the scikit-learn

```python
tfidf = TfidVectorizer(max_features=5000, stop_words='english')
X = tfidf.fit.transform(df['review'])
y = df['sentiment']
```
